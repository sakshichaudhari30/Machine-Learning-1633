**Q4. Data Preprocessing Prepare the dataset by:**
Separating features (X) and target (y)
Splitting the data into 80% training and 20% testing sets


```python
# Separate Features 

X = df_X
y = series_y
```

```python
# Splitting the data into 80 % training and 20% testing sets

from sklearn.model_selection import train_test_split
xtrain, xtest , ytrain , ytest = train_test_split(
    X, y,
    train_size= 0.8,
    random_state= 21
)
```

```python
print(xtrain.shape)
print(xtest.shape)
print(ytrain.shape)
print(ytest.shape)
```

**Q5. Model Training Build a Linear Regression model using the training dataset and display the model coefficients and intercept.**

```python
from sklearn.linear_model import LinearRegression

# Create Linear Regression model
model = LinearRegression()

# Train the model
model.fit(xtrain, ytrain)

# Display coefficients
print("Coefficients:", model.coef_)

# Display intercept
print("Intercept:", model.intercept_)
```

**Q6. Prediction**
Predict the target values for the testing dataset and display a comparison of the Actual and Predicted values

```python
ypred_test = model.predict(xtest)
ypred_test[:5]
```

```python
ytest.head()
```

**Q7. Model Evaluation Evaluate the model using:**
Mean Absolute Error (MAE)
Mean Squared Error (MSE)
Root Mean Squared Error (RMSE)
R² Score
Interpret the results.

```python

from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score

mse = mean_squared_error(ytest,ypred_test)
rmse = mse**(1/2)
mae = mean_absolute_error(ytest,ypred_test)
r2 = r2_score(ytest,ypred_test)

print(f"MSE:{mse}")
print(f"RMSE:{rmse}")
print(f"MAE:{mae}")
print(f"R2:{r2*100:.2f}%")
```
