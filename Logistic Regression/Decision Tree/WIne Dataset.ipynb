### Q3. Data Visualization
Create the following visualizations:
- Class Distribution Plot
- Correlation Heatmap
- Histogram of selected features
- Box Plot to identify outliers


Write your observations for each visualization.

```python
# Class Distribution plot
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
y['target'].value_counts().plot(kind='bar')

plt.title("Wine Class Distribution")
plt.xlabel("Class")
plt.ylabel("Count")
plt.show()
```

**Analysis**
- The dataset contains three wine classes: 0, 1, and 2.

- Class 1 is the most common.

- Class 2 is the least common.

```python
# Correlation Heatmap 
import seaborn as sns
import matplotlib.pyplot as plt
data = df_x.copy()
data["target"] = y
plt.figure(figsize=(20,15))
sns.heatmap(data.corr(),annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
# Histogram of 3 selected features

import matplotlib.pyplot as plt

features = ["flavanoids", "total_phenols", "od280/od315_of_diluted_wines"]

for feature in features:
    plt.hist(df_x[feature], bins=20, edgecolor="white")
    plt.xlabel(feature)
    plt.ylabel("Frequency")
    plt.title(f"Histogram of {feature}")
    plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df = pd.concat([df_x, y.rename(columns={y.columns[0]: "target"})], axis=1)

con = ["flavanoids", "total_phenols", "od280/od315_of_diluted_wines"]

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for i, col in enumerate(con):
    ax = axes[i]
    sns.boxplot(data=df, x="target", y=col, ax=ax, color="orange")
    ax.set_title(col)

plt.tight_layout()
plt.show()
```

**Analysis**

1. Flavanoids

- Class 0 has the highest median.

- Class 2 has the lowest median.

- Some outliers are present.

2. Total Phenols

- Class 0 has the highest median.

- Class 2 has the lowest median.

- A few outliers are present.

3. OD280/OD315 of Diluted Wines

- Class 0 has the highest median.

- Class 2 has the lowest median.

- Class 2 has a few outliers.

### Q4. Data Preprocessing

Prepare the dataset by:

- Separating features (X) and target (y)
- Splitting the data into 80% training and 20% testing sets using random_state=42
- Explain whether feature scaling is required for a Decision Tree Classifier.

```python
# Separate feature (X) and target(y)

X = df_x
y = y
```

```python
# Split the Dataset into 80% training and 20% testing 

from sklearn.model_selection import train_test_split
Xtrain,xtest,ytrain,ytest = train_test_split(
    X,y, train_size=0.8,random_state=42
)
print(Xtrain.shape)
print(xtest.shape)
print(ytrain.shape)
print(ytest.shape)
```

```python
#feature scaling is required for a Decision Tree Classifier.
```

#### Q5. Model Training
Build a Decision Tree Classifier using the training dataset and:
- Train the model
- Display the tree depth
- Display the number of leaf nodes
- Visualize the Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier

# Train the model using training data
model = DecisionTreeClassifier(random_state=42)
model.fit(Xtrain, ytrain)

# Display tree depth
print("Tree Depth:", model.get_depth())

# Display number of leaf nodes
print("Number of Leaf Nodes:", model.get_n_leaves())
```

```python
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree

plt.figure(figsize=(20, 10))

plot_tree(
    model,
    feature_names=Xtrain.columns,
    class_names=["Class 0", "Class 1", "Class 2"],
    filled=True,
    rounded=True
)

plt.show()
```
