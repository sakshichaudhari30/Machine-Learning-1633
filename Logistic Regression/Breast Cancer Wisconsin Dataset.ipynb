### Q4. Data Preprocessing
Prepare the dataset by:
Separating features (X) and target (y)
Splitting the data into 80% training and 20% testing sets

```python
# Separating fetures and target
X = df_x
y = y_series
```

```python
# Split into training and testing data

from sklearn.model_selection import train_test_split

xtrain, xtest,ytrain, ytest = train_test_split(X,y,train_size = 0.8,random_state=21)
print(xtrain.shape,xtest.shape)
print(ytrain.shape,ytest.shape)
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer

num_pipe = make_pipeline(SimpleImputer(strategy = 'median')).set_output(transform = "pandas")
num_pipe
```

```python
# fit it on training data
num_pipe.fit(xtrain)
```

```python
# Transform it on tesing as well as training data
xtrain_pre = num_pipe.transform(xtrain)
xtest_pre = num_pipe.transform(xtest)
xtrain_pre.head()
```

### Q5. Model Training
Build a Logistic Regression model using the training dataset and display the model coefficients and intercept.

```python
from sklearn.linear_model import LogisticRegression
```

```python
#Create model

model = LogisticRegression()
model.fit(xtrain_pre,ytrain)
```

```python
## model Coefficients
print("Coefficients:", model.coef_)
```

```python
## Display intercept 
print("Intercept :", model.intercept_)
```

### Q6. Prediction
Predict the target classes for the testing dataset and display a comparison of the Actual and Predicted values.

```python
ypred_test = model.predict(xtest)

ypred_test[:5]
ytest.head()
```

```python
# Comparing Actul and Predicted values
comparison = pd.DataFrame({
    "Actual":ytest.values,
    "Predicted" : ypred_test
})
comparison.head()
```

### Q7. Model Evaluation
Evaluate the model using:
- Accuracy Score
- Precision Score
- Recall Score
- F1-Score
- Confusion Matrix
- Classification Report
- Interpret the results.

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score
#prediction
ytest_preds = model.predict(xtest_pre)
#Accuracy 
accuracy_score = accuracy_score(ytest,ytest_preds)
accuracy_score
```

```python
#precision

precision_score =precision_score(ytest,ytest_preds)
precision_score
```

```python
# Recall 

recall_score = recall_score(ytest,ytest_preds)
recall_score
```

```python
# F1 Score
from sklearn.metrics import f1_score
f1 = f1_score(ytest,ytest_preds)
f1
```

```python
from sklearn.metrics import ConfusionMatrixDisplay, classification_report
# Confusion Matrix
ConfusionMatrixDisplay.from_estimator(model,xtest_pre,ytest)
```

```python
# Classification Report 
print(classification_report(ytest,ytest_preds))
```

#### Analysis
- Accuracy: 92% — most test samples were correctly classified.
- Class 0: Precision 92%, Recall 85%, F1-score 88%.
- Class 1: Precision 92%, Recall 96%, F1-score 94%.
- Macro F1-score: 91%.
- Overall: The model performed well, with higher recall for Class 1.
