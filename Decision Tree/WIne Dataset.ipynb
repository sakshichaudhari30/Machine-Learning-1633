### Q4. Data Preprocessing

Prepare the dataset by:

- Separating features (X) and target (y)
- Splitting the data into 80% training and 20% testing sets using random_state=42
- Explain whether feature scaling is required for a Decision Tree Classifier.

```python
# Separate feature (X) and target(y)

X = df_x
y = y_series
```

```python
# Split the Dataset into 80% training and 20% testing 

from sklearn.model_selection import train_test_split
Xtrain,xtest,ytrain,ytest = train_test_split(
    X,y, train_size=0.8,random_state=42
)
print(Xtrain.shape)
print(xtest.shape)
print(ytrain.shape)
print(ytest.shape)
```

```python
#feature scaling is required for a Decision Tree Classifier.
```

#### Q5. Model Training
Build a Decision Tree Classifier using the training dataset and:
- Train the model
- Display the tree depth
- Display the number of leaf nodes
- Visualize the Decision Tree

```python
# Train the model 

from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(random_state=42)
model.fit(Xtrain, ytrain)
```

```python
# Find the tree Depth 
model.get_depth()
```

```python
# number of leaf nodes 
model.get_n_leaves()
```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

plt.figure(figsize=(20, 10))

plot_tree(
    model,
    feature_names=Xtrain.columns,
    class_names=['Class 0', 'Class 1', 'Class 2'],
    filled=True,
    rounded=True,
    fontsize=8
)

plt.show()
```

### Q6. Prediction
Predict the target classes for the testing dataset and display a comparison of the Actual and Predicted values.

```python
# Predict target classes
y_pred = model.predict(xtest)

# Compare actual and predicted values
result = pd.DataFrame({
    'Actual': ytest,
    'Predicted': y_pred
})
print(result)
```

### Q7. Model Evaluation
Evaluate the model using:
           - Accuracy Score

           - Recall Score

```python
from sklearn.metrics import classification_report

y_pred = model.predict(xtest)

print(classification_report(ytest, y_pred))
```

### Q9. Hyperparameter Tuning

Train another Decision Tree model by changing parameters such as:

- max_depth
- criterion (gini or entropy)
- min_samples_split
Compare the performance of both models and discuss the results.

```python
params = {
    'max_depth': [3, 5, 6, 8, 10],
    'criterion': ['gini', 'entropy'],
    'min_samples_split': [2, 3, 4, 5, 6]
}
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.tree import DecisionTreeClassifier

base_model = DecisionTreeClassifier(random_state=42)

gscv = GridSearchCV(
    estimator=base_model,
    param_grid=params,
    cv=5,
    scoring='accuracy'
)

gscv.fit(Xtrain, ytrain)
```

```python
gscv.best_params_
```

```python
best_model = gscv.best_estimator_
```

```python
best_model
```

```python
from sklearn.metrics import accuracy_score

y_pred_original = model.predict(xtest)

print("Original Model Accuracy:",
      accuracy_score(ytest, y_pred_original))
```

```python
y_pred_tuned = best_model.predict(xtest)

print("Tuned Model Accuracy:",
      accuracy_score(ytest, y_pred_tuned))
```
